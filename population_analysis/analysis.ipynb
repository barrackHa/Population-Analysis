````python
# Create a single PSTH plot with all trial types and both directions
print("=== Single PSTH: All trial types and both directions ===")

# Color palette with distinct colors for each condition
colors = {
    ('GO', 0): '#2ca02c',       # Green for GO Right
    ('GO', 180): '#98df8a',     # Light green for GO Left
    ('STOP', 0): '#1f77b4',     # Blue for STOP Right
    ('STOP', 180): '#aec7e8',   # Light blue for STOP Left
    ('CONT', 0): '#ff7f0e',     # Orange for CONT Right
    ('CONT', 180): '#ffbb78',   # Light orange for CONT Left
}

# Create overlay of all conditions
all_psth = hv.NdOverlay()

for direction in [0, 180]:
    dir_label = "Right" if direction == 0 else "Left"
    
    for trial_type in ['GO', 'STOP', 'CONT']:
        # Get PSTH data
        bin_centers, firing_rate, n_trials = cell.calculate_psth(
            epok=[-200, 700],
            bin_size=10,
            alignment_point='go_cue',
            trial_type=trial_type,
            direction=direction,
            ssd_number=None,  # Combine all SSDs
            success_only=True,
            smooth=True
        )
        
        if bin_centers is not None:
            # Create curve with label
            label = f"{trial_type} {dir_label} (n={n_trials})"
            curve = hv.Curve((bin_centers, firing_rate), label=label).opts(
                color=colors[(trial_type, direction)],
                line_width=2
            )
            all_psth[label] = curve

# Add vertical line at time 0
zero_line = hv.VLine(0).opts(
    color='red', line_width=2, line_dash='dashed', alpha=0.7
)

# Combine and configure with legend
psth_all = (all_psth * zero_line).opts(
    opts.Curve(width=1000, height=500, show_grid=True, tools=['hover']),
    opts.NdOverlay(
        xlabel='Time from go_cue (ms)',
        ylabel='Firing Rate (spikes/s)',
        title=f'Cell {cell.cell_id} - All Trial Types and Directions (Success Only)\nPSTH with Gaussian Smoothing (bin=10ms)',
        legend_position='right',
        show_legend=True,
        legend_cols=1,
        legend_opts={'title': 'Conditions', 'label_text_font_size': '10pt'}
    ),
    opts.VLine(color='red', line_width=2, line_dash='dashed')
)

psth_all
````

# MSN Cell Analysis Framework - Summary

## Overview
This notebook implements a comprehensive analysis framework for Medium Spiny Neurons (MSN) recorded during a stop-signal task. The analysis replicates and extends the methodology from **Pani et al. (2022)** - "Neuronal Activity in the Primate Caudate Nucleus Relates to Visual Salience and Action Selection During Stop-Signal Task".

## Database Structure
- **Species**: Macaque monkey (Fiona)
- **Total trials**: 844,694 cell-trial combinations
- **Cells**: 1,587 MSN neurons
- **Sessions**: 59 recording sessions
- **Task**: Stop-signal paradigm with 3 trial types:
  - **GO**: No stop signal presented (baseline trials)
  - **STOP**: Stop signal presented, successful inhibition
  - **CONT**: Stop signal presented, failed inhibition (continue)
- **Directions**: 0° (right) and 180° (left) saccade targets
- **Stop Signal Delays (SSD)**: 4 levels (1-4), varying time between go and stop cues

## Key Classes

### 1. `Cell`
Represents a single neuron with all its trials. Core methods:

#### Spike Alignment
- `align_spikes_to_event(alignment_point)`: Align spike times to 'go_cue', 'stop_cue', or 'first_relevant_saccade'
- `filter_trials()`: Filter by type, direction, SSD, success/failure

#### Visualization Methods
- `plot_raster_by_type_direction()`: Raster plots with SSD color-coding and sorting
- `plot_histogram_by_type_direction()`: Spike count histograms (bin_size=10ms default)
- `plot_psth_by_type_direction()`: Peri-Stimulus Time Histograms with Gaussian smoothing

#### Data Processing
- `aggregate_spikes_by_bins()`: Bin spikes into temporal windows
- `calculate_psth()`: Compute firing rates with adaptive Gaussian smoothing (sigma=bin_size, truncate=2)

### 2. `PopulationAnalyzer`
Manages the full cell database and provides population-level analysis tools.

## Analysis Parameters

### Default Settings
- **Epoch**: [-200, 700] ms relative to alignment point
- **Bin size**: 10 ms
- **Smoothing**: Gaussian filter with sigma=bin_size, radius=2σ (±20ms for 10ms bins)
- **Trial selection**: Success-only trials (trial_failed=False) unless specified
- **Plot dimensions**: 800×300 px (individual), 1000×500 px (combined)

### Color Schemes
- **SSD-based**: Pink (SSD1), Brown (SSD2), Yellow-green (SSD3), Cyan (SSD4)
- **Type-based**: Green (GO), Blue (STOP), Orange (CONT)
- **Combined SSD plots**: Red (SSD1), Blue (SSD2), Green (SSD3), Orange (SSD4)
  - Solid lines: STOP trials (successful inhibition)
  - Dashed lines: CONT trials (failed inhibition)

## Usage Examples

### Basic Cell Analysis
```python
# Get a specific cell
cell = analyzer.get_cell(cell_id)

# Create rasters aligned to go cue
rasters = cell.plot_raster_by_type_direction(alignment_point='go_cue')
rasters[0]['STOP']  # Display STOP trials, right direction

# Generate PSTH with separate SSDs

psth[0]['STOP']  # Display PSTH for STOP trials, right direction
```

### Advanced Analysis
```python

psth_raw = cell.plot_psth_by_type_direction(smooth=False)
(psth_smooth[0]['STOP'] + psth_raw[0]['STOP']).cols(2)

    alignment_point='stop_cue',
    trial_type='STOP',

```

## Key Findings from Framework
This framework enables investigation of:
1. **Temporal dynamics**: How neural activity evolves relative to task events
2. **SSD effects**: Response differences across stop signal delays
3. **Success vs failure**: STOP (successful inhibition) vs CONT (failed inhibition) comparison
4. **Directional selectivity**: Left vs right saccade target differences
