```python
# Imports
import numpy as np
import matplotlib.pyplot as plt
import holoviews as hv
from sklearn.decomposition import PCA, TruncatedSVD

import pandas as pd

hv.extension('bokeh')

print("✓ Imports loaded successfully!")
```

```python
# Create 1000ms timeline with 1ms resolution
time_ms = np.arange(0, 1000, 1)  # 0 to 999 ms
n_timepoints = len(time_ms)

print(f"Timeline: {n_timepoints} timepoints from {time_ms[0]} to {time_ms[-1]} ms")
```

```python
# Generate three dimensions
# Convert time to seconds for sin calculation
time_sec = (2 * time_ms) / 1000.0

# Dimension 1: sin(t) - varies with time
# baseline = 4 
# get_left = lambda t: 2 / (1 + np.exp(-10 * (t - (np.random.normal(loc=0.75, scale=0.2))))) + baseline 
# get_right = lambda t: -2 / (1 + np.exp(-10 * (t - 0.75))) + baseline 

def get_psth(neuron_type, t, baseline=4):
    sgn = 1  # 1 for left, -1 for right
    if neuron_type == 'right':
        sgn = -1
    t0 = np.random.normal(loc=0.75, scale=0.25) 
    L = 2 + np.random.poisson(lam=0.5)
    k = 10
    return sgn * L / (1 + np.exp(-k * (t - t0))) + baseline

# left = 2 / (1 + np.exp(-10 * (time_sec - 0.75))) + baseline
# right = -2 / (1 + np.exp(-10 * (time_sec - 0.75))) + baseline

left = get_psth('left', time_sec)
right = get_psth('right', time_sec)

hv.NdOverlay({
    i: hv.Curve(d) for i,d in enumerate([left, right])
}).opts(
    width=800, height=300, tools=['hover'], 
    title=f"Neurons FR sim", xlabel="Time (ms)", ylabel="Value"
) 

```

```python
np.concat((left, right)).shape
```

```python
N = 50  # Number of neurons per type
psths_list = []
for i in range(N):
    psths_list.append(
        np.concat(
            (get_psth('left', time_sec),get_psth('right', time_sec))
        )
    )

data_matrix = np.stack(psths_list, axis=0)  # Shape: (neurons, timepoints)
data_matrix.shape  # Should be (100, 1000)
```

```python
pca = PCA(n_components=5)
pcs = pca.fit_transform(data_matrix.T)

# fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)
# for i in range(3):
#     axes[i].plot(time_ms, pcs[:, i], linewidth=2, c='brg'[i])
#     axes[i].set_ylabel(f'PC {i+1}', fontsize=12)
#     axes[i].grid(True, alpha=0.3)
```

```python

svd = TruncatedSVD(n_components=5)
evs = svd.fit(data_matrix.T)

svd_components = svd.transform(data_matrix.T)

# fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)
# for i in range(3):
#     axes[i].plot(time_ms, svd_components[:, i], linewidth=2, c='brg'[i])
#     axes[i].set_ylabel(f'EV {i+1}', fontsize=12)
#     axes[i].grid(True, alpha=0.3)
```

```python
fig, axes = plt.subplots(3, 2, figsize=(14, 8), sharex=True)
for i in range(3):
    axes[i,0].plot(time_ms, svd_components[:1000, i], linewidth=2, c='b')
    axes[i,0].plot(time_ms, svd_components[1000:, i], linewidth=2, c='r')
    axes[i,0].set_ylabel(f'EV {i+1}', fontsize=12)
    axes[i,0].grid(True, alpha=0.3)

    axes[i,1].plot(time_ms, pcs[:1000, i], linewidth=2, c='b')
    axes[i,1].plot(time_ms, pcs[1000:, i], linewidth=2, c='r')
    axes[i,1].set_ylabel(f'PC {i+1}', fontsize=12)
    axes[i,1].grid(True, alpha=0.3)

axes[0,0].set_title('SVD Components', fontsize=12, fontweight='bold')
axes[0,1].set_title('PCA Components', fontsize=12, fontweight='bold')

plt.tight_layout()
```
